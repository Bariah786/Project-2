```python
import heapq
from collections import Counter
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

def list_of_top_10_repeating_words(desc):
  #remove stopwords
  stop_words = set(stopwords.words('english'))
  desc_without_stopwords = [word for word in desc.split(' ') if word.lower() not in stop_words]
  cnt = Counter(desc_without_stopwords)
  top_k = [word for (word,_) in heapq.nlargest(10, cnt.items(), key=lambda x: x[1])]
  return top_k
```

```python
top_words_by_type = products_type.groupby('type')['desc'].sum().apply(list_of_top_10_repeating_words).reset_index()
top_words_by_type
```

```python
merged_df= top_words_by_type.merge(products_type, how= 'right', on = 'type')
```
